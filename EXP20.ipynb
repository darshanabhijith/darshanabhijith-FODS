```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_ind

# Load the dataset
df = pd.read_csv("/content/20_AB_Testing_Website_Dataset - 20_AB_Testing_Website_Dataset.csv")

# Display first five rows
print("First 5 rows of the dataset:")
print(df.head())

# Separate conversion data for Design A and Design B
design_A = df[df['Design'] == 'A']['Converted']
design_B = df[df['Design'] == 'B']['Converted']

# Calculate mean conversion rates
mean_A = design_A.mean()
mean_B = design_B.mean()

print("\nConversion Rate of Design A:", mean_A * 100, "%")
print("\nConversion Rate of Design B:", mean_B * 100, "%")

# Perform independent samples t-test
t_stat, p_value = ttest_ind(design_A, design_B, equal_var=False)

print("\nT-statistic:", t_stat)
print("P-value:", p_value)

# Check statistical significance
alpha = 0.05

if p_value < alpha:
    print("\nThere is a statistically significant difference between Design A and Design B.")
else:
    print("\nThere is no statistically significant difference between Design A and Design B.")
```
