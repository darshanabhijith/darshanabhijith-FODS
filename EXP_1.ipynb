```python
import numpy as np
import pandas as pd

# Load student scores from the selected CSV file
csv_path = "/content/student_scores - student_scores (2) - student_scores - student_scores (2).csv"
df = pd.read_csv(csv_path)

# Exclude the 'Student' column to get only the numeric scores
subjects = ["Math", "Science", "English", "History"]
student_scores = df[subjects].values

average = np.mean(student_scores, axis=0)

for i in range(len(subjects)):
    print(subjects[i], "Average:", round(average[i], 2))

highest = np.argmax(average)

print("Highest Average Subject:", subjects[highest])
```
