```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('/content/12.Temprain - 12.Temprain.csv')

print(data.head())
```

```python
data['Date'] = pd.to_datetime(
    data[['Year', 'Month']].assign(Day=1)
)

plt.figure(figsize=(12, 5))

plt.plot(data['Date'], data['tem'],
         color='red', label='Temperature')

plt.title('Monthly Temperature')
plt.xlabel('Year')
plt.ylabel('Temperature')
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.scatter(data['Date'], data['rain'],
            color='blue', label='Rainfall')

plt.title('Monthly Rainfall')
plt.xlabel('Year')
plt.ylabel('Rainfall')
plt.legend()
plt.grid(True)

plt.show()
```
