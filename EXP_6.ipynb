```python

# Grocery item prices and quantities
items = ["Rice", "Milk", "Bread", "Eggs", "Oil"]

prices = [60, 30, 40, 8, 150]
quantities = [5, 3, 2, 12, 2]

discount_rate = 10
tax_rate = 5

# Calculate total cost
total = 0

for i in range(len(items)):
    item_cost = prices[i] * quantities[i]
    total += item_cost

# Apply discount
discount = total * discount_rate / 100
discounted_total = total - discount

# Apply tax
tax = discounted_total * tax_rate / 100
final_amount = discounted_total + tax

print("Original Total: Rs.", total)
print("Discount: Rs.", discount)
print("Tax: Rs.", tax)
print("Final Amount: Rs.", final_amount)
```
