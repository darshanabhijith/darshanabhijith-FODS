```python
import pandas as pd

# Correct path based on actual environment files
file_path = '/content/Sales (1).xlsx - Sales.csv'
df = pd.read_csv(file_path)

# Display columns and head
print(df.info())
display(df.head())
```

```python
# Calculate item-level cost: Price * Quantity_Sold
df['Calculated_Total'] = df['Price'] * df['Quantity_Sold']

# Overall original total
total = df['Calculated_Total'].sum()

# Average discount applied in the dataset or a standard discount
# Let's apply a 10% discount to keep it consistent with your original logic
discount_rate = 10
tax_rate = 5

discount = total * discount_rate / 100
discounted_total = total - discount
tax = discounted_total * tax_rate / 100
final_amount = discounted_total + tax

print("--- Calculation Results from Dataset ---")
print(f"Original Total: Rs. {total:,.2f}")
print(f"Discount (10%): Rs. {discount:,.2f}")
print(f"Tax (5%): Rs. {tax:,.2f}")
print(f"Final Amount: Rs. {final_amount:,.2f}")
```
