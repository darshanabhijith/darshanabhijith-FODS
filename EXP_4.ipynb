```python

import numpy as np

# Quarterly sales from the dataset
sales_data = np.array([
    1247650.49,
    1246529.71,
    1185251.76,
    1320505.26
])

# Calculate total annual sales
total_sales = np.sum(sales_data)

# Calculate percentage increase
percentage_increase = (
    (sales_data[3] - sales_data[0]) / sales_data[0]
) * 100

print("Quarterly Sales:", sales_data)
print("Total Annual Sales:", round(total_sales, 2))
print("Percentage Increase:", round(percentage_increase, 2), "%")
```
