```python
import pandas as pd
import numpy as np

# Load the CSV file
file_path = '/content/Sales (1).xlsx - Sales.csv'
df = pd.read_csv(file_path)

# Display the dataset info and preview the structure
display(df.head())
print(df.info())
```

```python
import pandas as pd
import numpy as np

# Ensure 'Month_sales' is parsed as datetime
df['Month_sales_dt'] = pd.to_datetime(df['Month_sales'], format='%B %Y')

# Filter dataset to include only sales from the year 2023
df_2023 = df[df['Month_sales_dt'].dt.year == 2023].copy()

# Map each record to its corresponding quarter (e.g., 2023Q1, 2023Q2, etc.)
df_2023['Quarter'] = df_2023['Month_sales_dt'].dt.to_period('Q')

# Group by Quarter and sum the 'Total' sales column
quarterly_sales_series = df_2023.groupby('Quarter')['Total'].sum().sort_index()

sales_by_quarter = quarterly_sales_series.values
quarters = quarterly_sales_series.index.astype(str).tolist()

# Calculate total annual sales for 2023	otal_sales = np.sum(sales_by_quarter)

# Calculate percentage increase from Q1 to Q4 for 2023
percentage_increase = ((sales_by_quarter[3] - sales_by_quarter[0]) / sales_by_quarter[0]) * 100

# Print structured results
for q, val in zip(quarters, sales_by_quarter):
    print(f"{q} Sales: ${val:,.2f}")

print("\nTotal Annual Sales (2023): ${:,.2f}".format(total_sales))
print("Percentage Increase (2023 Q1 to Q4): {:.2f}%".format(percentage_increase))
```
