```python

import numpy as np

# Columns: Price, Area, Bedrooms
house_data = np.array([
    [10150000, 16200, 5],
    [8400000, 7950, 5],
    [8120000, 6840, 5],
    [6440000, 8580, 5],
    [6083000, 4300, 6],
    [5600000, 4800, 5],
    [5565000, 5400, 5],
    [4340000, 1905, 5],
    [3850000, 5300, 5],
    [3773000, 2520, 5],
    [3500000, 3600, 6],
    [1960000, 3420, 5]
])

bedrooms = house_data[:, 2]
prices = house_data[:, 0]

selected_prices = prices[bedrooms > 4]

average_price = np.mean(selected_prices)

print("Houses with more than 4 bedrooms:", len(selected_prices))
print("Average Sale Price:", round(average_price, 2))
```
