```python
import numpy as np

# Loading 'price' (col 0) and 'bedrooms' (col 2) from your specific CSV file
# Using skip_header=1 for np.genfromtxt to skip the header row correctly
house_data = np.genfromtxt('/content/Exp9 - Exp9.csv', delimiter=',', skip_header=1, usecols=(0, 2))

# Index 1 of our loaded array represents the 'bedrooms' column because we only selected columns 0 and 2
houses_more_than_4 = house_data[house_data[:, 1] > 4]

# Index 0 of our loaded array represents the 'price' column
average_price = np.mean(houses_more_than_4[:, 0])

print("Average price of houses with more than 4 bedrooms: ", round(average_price, 2))
```
