```python
import pandas as pd

data = {
    'Customer_ID': [1, 2, 3, 4, 5, 6, 7, 8],
    'Age': [22, 25, 22, 30, 25, 35, 30, 22]
}

df = pd.DataFrame(data)

print(df)
```

```python
age_frequency = df['Age'].value_counts().sort_index()

print("Frequency Distribution of Customer Ages:")
print(age_frequency)
```

```python
import matplotlib.pyplot as plt
```

```python
age_frequency.plot(kind='bar', color='skyblue')

plt.title('Frequency Distribution of Customer Ages')
plt.xlabel('Customer Age')
plt.ylabel('Number of Customers')
plt.show()
```
