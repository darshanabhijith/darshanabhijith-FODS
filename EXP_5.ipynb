```python
import pandas as pd
import numpy as np

# Path to the selected dataset
file_path = '/content/Fuel_Efficiency datasets.xlsx - Sheet1 (1) - Fuel_Efficiency datasets.xlsx - Sheet1 (1).csv'

# Load dataset
df = pd.read_csv(file_path)

# Display preview of dataset to locate fuel efficiency column
display(df.head())
print("Columns:", df.columns.tolist())
```

```python
# Explicitly use the 'combination_mpg' column for combined fuel efficiency
fuel_efficiency = df['combination_mpg'].dropna().values

average = np.mean(fuel_efficiency)

ford_mustang = 22
mazda_6 = 29
percentage = ((mazda_6 - ford_mustang) / ford_mustang) * 100

print("Number of Vehicles:", len(fuel_efficiency))
print("Average Fuel Efficiency:", round(average, 2), "MPG/MPGe")
print("Percentage Improvement:", round(percentage, 2), "%")
```
