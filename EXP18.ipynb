```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("18bodyfat FINAL - 18bodyfat FINAL.csv")

print("Mean Age:", round(df['Age'].mean(), 2))
print("Mean Body Fat:", round(df['BodyFat'].mean(), 2))

print("Median Age:", df['Age'].median())
print("Median Body Fat:", df['BodyFat'].median())

print("Standard Deviation of Age:", round(df['Age'].std(), 2))
print("Standard Deviation of Body Fat:", round(df['BodyFat'].std(), 2))

plt.boxplot([df['Age'], df['BodyFat']],
            labels=['Age', 'Body Fat'])
plt.title("Distribution of Age and Body Fat")
plt.show()

plt.scatter(df['Age'], df['BodyFat'], marker='o')
plt.title("Age vs Body Fat")
plt.xlabel("Age")
plt.ylabel("Body Fat")
plt.show()
```
