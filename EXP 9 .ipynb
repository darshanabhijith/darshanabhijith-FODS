```python

import pandas as pd

property_data = pd.read_csv("/content/Exp9 - Exp9.csv")

print("First 5 Records:")
print(property_data.head())

# 1. Average listing price by location
if "location" in property_data.columns:
    print("\nAverage Price by Location:")
    print(property_data.groupby("location")["price"].mean())
else:
    print("\nLocation column is not available in the dataset.")

# 2. Properties with more than 4 bedrooms
count = (property_data["bedrooms"] > 4).sum()
print("\nProperties with more than 4 bedrooms:", count)

# 3. Property with the largest area
largest = property_data.loc[property_data["area"].idxmax()]

print("\nProperty with Largest Area:")
print(largest[["price", "area", "bedrooms"]])
```
