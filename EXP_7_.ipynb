```python
import pandas as pd

# Load the dataset from the selected CSV file
file_path = '/content/Exp7 - Exp7.csv'
order_data = pd.read_csv(file_path)

# Ensure order_date is treated as datetime
# We will let pandas automatically parse, or specify dayfirst=True if needed based on the format
order_data['order_date'] = pd.to_datetime(order_data['order_date'], dayfirst=True, errors='coerce')

# 1. Total number of distinct orders per customer
orders_per_customer = order_data.groupby('customer_name')['order_id'].nunique()

# 2. Average ordered quantity for each product
avg_qty_per_product = order_data.groupby('product_name')['quantity'].mean()

# 3. Earliest and latest dates
earliest_date = order_data['order_date'].min()
latest_date = order_data['order_date'].max()

print('Total Records:', len(order_data))
print('\nOrders per Customer:\n', orders_per_customer.to_string())
print('\nAverage Order Quantity by Product:\n', avg_qty_per_product.to_string())
if pd.notnull(earliest_date):
    print('\nEarliest Order Date:', earliest_date.strftime('%d-%m-%Y'))
else:
    print('\nEarliest Order Date: N/A')
if pd.notnull(latest_date):
    print('Latest Order Date:', latest_date.strftime('%d-%m-%Y'))
else:
    print('Latest Order Date: N/A')
```
