# Program 34: Patient Medical Treatment Outcomes Model (KNN)

**Scenario / Problem Statement**

Predict treatment outcome (Good or Bad) using patient age, blood pressure, and cholesterol measurements with a K-Nearest Neighbors classifier.

Run the cells from top to bottom in Google Colab.

> **Sample inputs used for this run:** new_age = 45, new_bp = 125, new_chol = 200. Values are labeled in the code.

## 1. Import required libraries

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import classification_report, accuracy_score
```

## 2. Load data and perform the analysis

```python
# Load the supplied CSV. In Colab, upload it if it is not already present.
from pathlib import Path
DATA_FILE = 'FODS_Lab_Datasets_Q20_to_Q40.xlsx - Q34_Medical_Treatment_KNN.csv'
if not Path(DATA_FILE).exists():
    try:
        from google.colab import files
        files.upload()
    except ImportError as exc:
        raise FileNotFoundError('FODS_Lab_Datasets_Q20_to_Q40.xlsx - Q34_Medical_Treatment_KNN.csv is missing; place the CSV beside this notebook.') from exc
df_raw = pd.read_csv(DATA_FILE, skiprows=3)
print("Dataset loaded:", DATA_FILE)
print("Dataset shape:", df_raw.shape)
display(df_raw.head())

features = ["Age", "Blood_Pressure", "Cholesterol"]
X = df_raw[features].apply(pd.to_numeric, errors="coerce")
y = df_raw["Treatment_Outcome"].astype(str).str.strip()
valid = X.notna().all(axis=1) & y.notna()
X, y = X.loc[valid], y.loc[valid]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)
knn = make_pipeline(StandardScaler(), KNeighborsClassifier(n_neighbors=5))
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

print(f"Overall Classification Accuracy: {accuracy_score(y_test, y_pred):.4f}\n")
print(classification_report(y_test, y_pred))

new_age = 45  # SAMPLE INPUT
new_bp = 125  # SAMPLE INPUT
new_chol = 200  # SAMPLE INPUT
new_patient = pd.DataFrame([[new_age, new_bp, new_chol]], columns=features)
print("\nPredicted Treatment Outcome:", knn.predict(new_patient)[0])
```
