# Program 22: Customer Reviews & Average Rating Confidence Intervals

**Scenario / Problem Statement**

Load customer ratings, calculate the average rating, and output the 95% confidence interval for the true population mean.

Run the cells from top to bottom in Google Colab.

## 1. Import required libraries

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
```

## 2. Upload and load the supplied dataset

```python
# Load the supplied CSV. In Colab, upload it if it is not already present.
from pathlib import Path
DATA_FILE = 'FODS_Lab_Datasets_Q20_to_Q40.xlsx - Q22_Customer_Reviews.csv'
if not Path(DATA_FILE).exists():
    try:
        from google.colab import files
        files.upload()
    except ImportError as exc:
        raise FileNotFoundError('FODS_Lab_Datasets_Q20_to_Q40.xlsx - Q22_Customer_Reviews.csv is missing; place the CSV beside this notebook.') from exc
df_raw = pd.read_csv(DATA_FILE, skiprows=3)
print("Dataset loaded:", DATA_FILE)
print("Shape:", df_raw.shape)
display(df_raw.head())

ratings = pd.to_numeric(df_raw["Rating"], errors="coerce").dropna()
mean_rating = ratings.mean()
sem_rating = stats.sem(ratings)
ci_rating = stats.t.interval(0.95, df=len(ratings)-1, loc=mean_rating, scale=sem_rating)

print(f"Total Reviews Analyzed: {len(ratings)}")
print(f"Mean Rating: {mean_rating:.2f}")
print(f"95% Confidence Interval for True Mean Rating: ({ci_rating[0]:.4f}, {ci_rating[1]:.4f})")
```
