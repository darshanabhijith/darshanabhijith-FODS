# Program 21: Rare Elements Concentration & Precision Estimation

**Scenario / Problem Statement**

Load rare-element concentration measurements, accept a sample size and confidence level, and calculate the sample point estimate and t confidence interval.

Run the cells from top to bottom in Google Colab.

> **Sample inputs used for this run:** sample_size = min(30, len(df_raw)), conf_level = 0.95. Values are labeled in the code.

## 1. Import required libraries

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
```

## 2. Upload and load the supplied dataset

```python
# Load the supplied CSV. In Colab, upload it if it is not already present.
from pathlib import Path
DATA_FILE = 'FODS_Lab_Datasets_Q20_to_Q40.xlsx - Q21_Rare_Elements.csv'
if not Path(DATA_FILE).exists():
    try:
        from google.colab import files
        files.upload()
    except ImportError as exc:
        raise FileNotFoundError('FODS_Lab_Datasets_Q20_to_Q40.xlsx - Q21_Rare_Elements.csv is missing; place the CSV beside this notebook.') from exc
df_raw = pd.read_csv(DATA_FILE, skiprows=3)
print("Dataset loaded:", DATA_FILE)
print("Shape:", df_raw.shape)
display(df_raw.head())

# Sample size and confidence level
sample_size = min(30, len(df_raw))  # SAMPLE INPUT
conf_level = 0.95  # SAMPLE INPUT

if not 1 <= sample_size <= len(df_raw):
    raise ValueError("Sample size must be between 1 and the number of records.")
if not 0 < conf_level < 1:
    raise ValueError("Confidence level must be between 0 and 1.")

sample_data = df_raw["Concentration_ppm"].dropna().sample(n=sample_size, random_state=42)
point_estimate = sample_data.mean()
sem = stats.sem(sample_data)
ci = stats.t.interval(conf_level, df=sample_size-1, loc=point_estimate, scale=sem)

print(f"\nSample Size: {sample_size}")
print(f"Point Estimate of Mean Concentration: {point_estimate:.4f} ppm")
print(f"{conf_level*100:.1f}% Confidence Interval: ({ci[0]:.4f}, {ci[1]:.4f}) ppm")
```
