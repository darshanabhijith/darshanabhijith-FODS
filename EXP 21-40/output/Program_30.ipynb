# Program 30: CART (Decision Tree Regressor) for Car Price Prediction

**Scenario / Problem Statement**

Implement a Decision Tree Regressor to predict pre-owned car prices based on mileage and vehicle age.

Run the cells from top to bottom in Google Colab.

> **Sample inputs used for this run:** in_mileage = 40000, in_age = 4. Values are labeled in the code.

## 1. Import required libraries

```python
import numpy as np
import pandas as pd
from sklearn.tree import DecisionTreeRegressor, export_text
```

## 2. Prepare data and run the model

```python
# Generate 500 car listing records (as in the supplied program)
np.random.seed(30)
mileage = np.random.randint(5000, 150000, size=500)
age = np.random.randint(1, 15, size=500)
price = 35000 - age * 1800 - mileage * 0.1 + np.random.normal(0, 2000, size=500)

X = pd.DataFrame({"Mileage": mileage, "Age": age})
y = pd.Series(price)

reg = DecisionTreeRegressor(max_depth=3, random_state=42)
reg.fit(X, y)

in_mileage = 40000  # SAMPLE INPUT
in_age = 4  # SAMPLE INPUT

new_car = pd.DataFrame([[in_mileage, in_age]], columns=["Mileage", "Age"])
pred_val = reg.predict(new_car)[0]
print(f"\nPredicted Car Price: ${pred_val:,.2f}")

print("\nDecision Path Structure:")
print(export_text(reg, feature_names=["Mileage", "Age"]))
```
