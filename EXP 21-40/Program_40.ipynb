## 1. Import required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Load data and perform the analysis

```python
# Generate 500 custom soccer player records (as in the supplied program)
np.random.seed(40)
positions = ["Forward", "Midfielder", "Defender", "Goalkeeper"]
names = [f"Player_{i}" for i in range(1, 501)]
ages = np.random.randint(18, 38, size=500)
assigned_positions = np.random.choice(positions, p=[0.25, 0.35, 0.30, 0.10], size=500)
goals = np.where(assigned_positions == "Forward", np.random.randint(5, 30, size=500), np.random.randint(0, 8, size=500))
salaries = np.random.randint(10000, 250000, size=500)

df_soccer = pd.DataFrame({
    "Name": names,
    "Age": ages,
    "Position": assigned_positions,
    "Goals": goals,
    "Weekly_Salary": salaries
})

print("Top 5 Goal Scorers:")
display(df_soccer.nlargest(5, "Goals")[["Name", "Goals", "Position"]])

print("Top 5 Highest Weekly Salaries:")
display(df_soccer.nlargest(5, "Weekly_Salary")[["Name", "Weekly_Salary", "Position"]])

position_counts = df_soccer["Position"].value_counts()
plt.figure(figsize=(8, 5))
position_counts.plot(kind="bar")
plt.title("Soccer Players by Position")
plt.xlabel("Position")
plt.ylabel("Number of Players")
plt.xticks(rotation=0)
plt.grid(axis="y")
plt.show()
```
