**Scenario / Problem Statement**

Build a multiple linear regression model to predict car prices using engine size and horsepower.

Run the cells from top to bottom in Google Colab.

## 1. Import required libraries

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
```

## 2. Load data and perform the analysis

```python
# Upload the exact CSV dataset when prompted
from google.colab import files
uploaded = files.upload()

DATA_FILE = 'FODS_Lab_Datasets_Q20_to_Q40.xlsx - Q33_Car_Price_MultiReg(1).csv'
df_raw = pd.read_csv(DATA_FILE, skiprows=3)
print("Dataset loaded:", DATA_FILE)
print("Dataset shape:", df_raw.shape)
display(df_raw.head())

for col in ["Engine_Size_L", "Horsepower", "Car_Price_USD"]:
    df_raw[col] = pd.to_numeric(df_raw[col].astype(str).str.replace("$", "", regex=False).str.replace(",", "", regex=False), errors="coerce")
data = df_raw.dropna(subset=["Engine_Size_L", "Horsepower", "Car_Price_USD"])
X = data[["Engine_Size_L", "Horsepower"]]
y = data["Car_Price_USD"]

model = LinearRegression()
model.fit(X, y)
preds = model.predict(X)

print("Feature Coefficients:")
for col, coef in zip(X.columns, model.coef_):
    print(f" {col}: {coef:.2f}")
print(f"Intercept: {model.intercept_:.2f}")
print(f"R-squared value: {r2_score(y, preds):.4f}")
print(f"RMSE: {np.sqrt(mean_squared_error(y, preds)):,.2f}")
```
