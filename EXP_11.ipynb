```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv('/content/2,4,11Sales (1).xlsx - Sales.csv')

print(data.head())
```

```python
monthly_sales = data.groupby('Month_sales')['Total'].sum()

monthly_sales.index = pd.to_datetime(
    monthly_sales.index, format='%B %Y'
)

monthly_sales = monthly_sales.sort_index()

print(monthly_sales)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(monthly_sales.index, monthly_sales.values,
         marker='o', linestyle='-', label='Monthly Sales')

plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(monthly_sales.index, monthly_sales.values,
            color='red', label='Monthly Sales')

plt.title('Monthly Sales Scatter Plot')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(monthly_sales.index.strftime('%b %Y'),
        monthly_sales.values, color='skyblue')

plt.title('Monthly Sales Bar Chart')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```
