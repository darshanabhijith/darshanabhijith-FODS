```python

import pandas as pd
import numpy as np

df = pd.read_csv("19_blood_pressure_trial_500_patients - 19_blood_pressure_trial_500_patients.csv")

drug = df[df['Group'] == 'Drug']['Reduction']
placebo = df[df['Group'] == 'Placebo']['Reduction']

mean_drug = drug.mean()
mean_placebo = placebo.mean()

se_drug = drug.std() / np.sqrt(drug.count())
se_placebo = placebo.std() / np.sqrt(placebo.count())

z = 1.96

ci_drug = (mean_drug - z * se_drug,
           mean_drug + z * se_drug)

ci_placebo = (mean_placebo - z * se_placebo,
              mean_placebo + z * se_placebo)

print("Drug Group")
print("Mean BP Reduction:", round(mean_drug, 2))
print("95% Confidence Interval:",
      (round(ci_drug[0], 2), round(ci_drug[1], 2)))

print("\nPlacebo Group")
print("Mean BP Reduction:", round(mean_placebo, 2))
print("95% Confidence Interval:",
      (round(ci_placebo[0], 2), round(ci_placebo[1], 2)))
```
