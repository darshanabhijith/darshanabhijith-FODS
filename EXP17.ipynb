```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/17customer_reviews.csv', encoding='latin-1')

# The dataset uses 'Review Text' instead of 'reviewText'
review = df['Review Text'].dropna().str.lower().str.split().explode().value_counts()
remove = ["the","and","is"]
rem_w = review[~review.index.isin(remove)]
print(rem_w)
print(rem_w.nlargest(5))

review = rem_w.nlargest(5)
plt.bar(review.index, review.values)
plt.xlabel("Words")
plt.ylabel("Frequency")
plt.title("Most Frequent Words")
plt.show()
```
